```python
import numpy as np

marks = np.array([10,30,40,60,90])
hours = np.array([1,1.5,2,3,4])
w = 0.0
b = 0.0
learning_rate = 0.01

for epoch in range(1000):
    y_cap = w * hours + b
    error = y_cap - marks
    loss = np.mean(error**2)

    dw = (2/len(hours))*np.sum(error*hours)
    db = (2/len(hours))*np.sum(error)

    w = w - learning_rate * dw
    b = b - learning_rate * db

#Prediction
new_data = 1
predict = w * new_data + b

if(predict >= 100):
    print('Excellent: ',predict)
else:
    print(predict)
```
