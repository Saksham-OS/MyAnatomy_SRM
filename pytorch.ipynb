```python
a = torch.tensor(5.0 , requires_grad = True)
b = torch.tensor(8.0 , requires_grad = True)
ab = a * b
print(ab)
ab.backward()
print('Gradient of a:',float(a.grad))
print('Gradient of b:',float(b.grad))
```
