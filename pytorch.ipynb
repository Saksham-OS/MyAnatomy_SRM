Autograd

```python
w = torch.tensor(5.0, requires_grad = True)
b = torch.tensor(8.0, requires_grad = True)
optimiser = torch.optim.SGD([w,b], lr = 0.01)

marks = torch.tensor([10,30,40,60,80])
hours = torch.tensor([0.50, 1,1.5,2,3])

for epochs in range(100):
    optimiser.zero_grad()
    y_cap = w * hours + b

    loss = ((y_cap - marks) ** 2).mean()
    loss.backward()
    optimiser.step()

print('Weight is: ',w.item())
print('Bias is: ',b.item())
```

nn.module

```python
import torch.nn as nn
x = torch.randint(0,30,[4,4]).float()
y = torch.randint(0,30,[4,4]).float()

class LinearModel(nn.Module):
    def __init__(self):
        super().__init__()
        self.layer = nn.Linear(4,4) #(1,1) means one input and one output
    def forward(self, x):
        return self.layer(x)

model = LinearModel()
loss_function = nn.MSELoss() #Calculates loss ((y_cap - y)**2).mean()
optimiser = torch.optim.SGD(model.parameters(), lr = 0.0001)

for epoch in range(1000):
    y_cap = model(x) #y_cap = w * x + b and it also calls forward function.
    loss = loss_function(y_cap, y) #Calculates loss ((prediction - actual_value)**2).mean()
    optimiser.zero_grad() #Set the previous gradient to zero.
    loss.backward() #It calculates gradients/derivatives
    optimiser.step() #Updates parameters (weights and bias) using the formula (w_new = w_old - learning_rate * gradient)

print("Weight: ",model.layer.weight)
print("Bias: ",model.layer.bias)
```
